```python
import geopandas as gpd
import pandas as pd
import re
```

Read in the DeFlock data, ensure each is a point and make columns, reduce to only the necessary columns

```python
alpr = gpd.read_file("data/raw/alpr.geojson")
alpr = alpr[alpr.geometry.geom_type == "Point"].reset_index()
keep = ["id", "geometry", "operator", "surveillance:zone"]
alpr = alpr[[i for i in keep if i in alpr.columns]]
alpr.head(20)
```

Read in the traffic data with only the necessary columns and output the unique jurisdictions
